```python
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/e/2PACX-1vTajM7yamJD2vwSUSxFciRmydE1anE8d9Eni_crfuWOCrxM-bkm2FKF_wIUmRhHh58eciM7dmQCR8Or/pub?gid=0&single=true&output=csv"
df = pd.read_csv(url)
df = df.dropna(axis=1, how="all")

print(df)

# ¿Cuánto le he prestado en total a cada persona? (por si alguien pide prestado más de una vez)
por_persona = df.groupby("name")["amount"].sum()
print(por_persona)

# ¿Cuánto dinero tengo repartido en préstamos en este momento, con todo e interés?
deuda_total = df["total"].sum()
print(f"Total pendiente de cobrar (con interés): ${deuda_total}")

# ¿Cuánta ganancia esperada tienes en total si todos pagan?
ganancia_total = df["profit"].sum()
print(f"Ganancia esperada total: ${ganancia_total}")
```
